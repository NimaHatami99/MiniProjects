```python
import cv2 as cv 
import numpy as np

def find_water_pixels(input_image):

    # Creating an output mask image (grayscale)
    output_image = np.zeros(input_image.shape[:2], dtype=np.uint8)

    # Defining the threshold and ratio
    threshold = 150
    ratio = 1.05

    # Iterating through the image pixels and apply the conditions
    for i in range(input_image.shape[0]):
        for j in range(input_image.shape[1]):
            b, g, r = input_image[i, j]

            # Condition 1: All RGB values are more than 150 or R > G and R > B
            if (r > threshold and g > threshold and b > threshold) or (r > g and r > b):
                output_image[i, j] = 0  
            # Condition 2: B > G and B > R
            elif b > g and b > r:
                output_image[i, j] = 255  
            # Condition 3: G > R and G > B, and G < B * 1.05
            elif g > r and g > b and g < b * ratio:
                output_image[i, j] = 255  
    
    return output_image
```

```python
# Loading the images
image_1984 = cv.imread('دریاچه آرال_1984.png',cv.IMREAD_COLOR)
image_2022 = cv.imread('دریاچه آرال_2022.png',cv.IMREAD_COLOR)

# Resizing 1984 image
image_1984 = cv.resize(image_1984, (image_2022.shape[1], image_2022.shape[0]))

# Using the function 
thresh_1984 = find_water_pixels(image_1984)
thresh_2022 = find_water_pixels(image_2022)

# Calculating the dried area in 2022
total_lake_area_2022 = cv.countNonZero(thresh_2022)

# Calculating the total lake area in 1984
total_lake_area_1984 = cv.countNonZero(thresh_1984)

# Calculating the percentage of dried area
percentage_dried = (1 - (total_lake_area_2022 / total_lake_area_1984)) * 100

print(f"Percentage of lake dried up from 1984 to 2022: {percentage_dried:.2f}%") 

# Saving images 
cv.imwrite('resized_1984.jpg', image_1984)
cv.imwrite('thresh_1984.jpg', thresh_1984)
cv.imwrite('thresh_2022.jpg', thresh_2022)

# Displaying images and binary masks 
cv.imshow('image 1984', image_1984)
cv.imshow('image 2022', image_2022)
cv.imshow('mask image 1984', thresh_1984)
cv.imshow('mask image 2022', thresh_2022)
cv.waitKey(0)
cv.destroyAllWindows()
```
