```python
import cv2 as cv 
import numpy as np 
import zipfile 

# Extracting images from the ZIP file and store them in an array
zip_file_name = 'foyr_way_autobahn.zip'
image_folder_name = 'foyr_way_autobahn'

images = []
with zipfile.ZipFile(zip_file_name, 'r') as zip_file:
    for i in range(36):  # there are 36 images (street0.jpg to street35.jpg)
        image_name = f'{image_folder_name}/street{i}.jpg'
        with zip_file.open(image_name) as file:
            image_data = file.read()
            image_np = np.asarray(bytearray(image_data), dtype=np.uint8)
            image = cv.imdecode(image_np, cv.IMREAD_COLOR)
            images.append(image)

for method in ['average', 'median']:
    for num in [5, 10, 20, 35]:
        # Creating a list of frames for processing
        frames_to_process = images[0 : num]

        # Calculating the average or median frame
        if method == 'average':
            result_frame = np.mean(frames_to_process, axis=0).astype(np.uint8)
        elif method == 'median':
            result_frame = np.median(frames_to_process, axis=0).astype(np.uint8)

        # Displaying the resulting frame
        cv.imshow(f'{method} {num} frame', result_frame)
        cv.imwrite(f'{method}_{num}.jpg', result_frame)

cv.waitKey(0)
cv.destroyAllWindows()
```
