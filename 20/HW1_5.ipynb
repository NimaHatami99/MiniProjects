```python
#____________________________ part a ______________________________#
from sklearn import datasets
from sklearn.model_selection import train_test_split 

iris_data=datasets.load_iris()
iris_target=iris_data['target']
iris_values=iris_data.data[:, :4]
val_train, val_test, target_train, target_test=train_test_split(iris_values, iris_target, test_size=0.2, random_state=42)
print("training values' matrix has {} shape.".format(val_train.shape))
print("training targets' vector has {} shape.".format(target_train.shape))
```

```python
#____________________________ part b ______________________________#
from sklearn.preprocessing import StandardScaler

iris_scaler=StandardScaler() 
val_test=iris_scaler.fit_transform(val_test) 
val_train=iris_scaler.fit_transform(val_train) 
print("scaled test and train values have a mean of {} and {} respectively.".format(val_test.mean(),val_train.mean()))
print("scaled test and train values have a standard deviation of {} and {} respectively.".format(val_test.std(),val_train.std()))
```

```python
#____________________________ part c ______________________________#
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics

k_list=list(range(1,21))
acc_list=[]
for i in k_list:
    iris_knn=KNeighborsClassifier(n_neighbors=i)
    iris_knn.fit(val_train,target_train)
    target_pred=iris_knn.predict(val_test)
    acc_list.append(metrics.accuracy_score(target_test,target_pred))
optimal_k_list=[]
for j in k_list:
    if acc_list[j-1]==max(acc_list):
        optimal_k_list.append(j)
print("the optimal value for k can be any of the values in matrix below which gives a maximum accuracy of {:.2f}% on test samples.\n".format(max(acc_list)*100),optimal_k_list)
```
