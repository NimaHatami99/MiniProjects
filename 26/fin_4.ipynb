```python
import cv2 as cv

def click(event, y, x, flags, params):
    if event == cv.EVENT_LBUTTONDOWN:
        seed = []
        position = x, y
        seed.append(position)
        for pix in seed:
            if neighbor_type:
                iteratable = [[pix[0]+1, pix[1]], [pix[0]-1, pix[1]], [pix[0], pix[1]+1], 
                              [pix[0], pix[1]-1], [pix[0]-1, pix[1]-1], [pix[0]-1, pix[1]+1], 
                              [pix[0]+1, pix[1]-1], [pix[0]+1, pix[1]+1]]
            else:
                iteratable = [[pix[0]+1, pix[1]], [pix[0]-1, pix[1]], [pix[0], pix[1]+1], [pix[0], pix[1]-1]]
            for iteratable_pix in iteratable:
                diff = abs(int(img[seed[0][0], seed[0][1]]) - int(img[iteratable_pix[0], iteratable_pix[1]]))
                if diff < 11:
                    if tuple(iteratable_pix) not in seed:
                        seed.append(tuple(iteratable_pix))
                        img[iteratable_pix[0], iteratable_pix[1]] = 255
                        cv.imshow('input', img)
                        cv.waitKey(1)

        if neighbor_type:
            cv.imwrite('out8.png', img)
        else:
            cv.imwrite('out4.png', img)

operator_input=str(input("enter the method to show. out4 or out8?"))
if operator_input=='out4':
    neighbor_type = False
elif operator_input=='out8':
    neighbor_type = True

img = cv.imread('fMRI.jpg', cv.IMREAD_GRAYSCALE)
cv.imshow('input', img)
cv.namedWindow('input')
print ('please click on image')
cv.setMouseCallback('input', click)
cv.waitKey(0)
cv.destroyAllWindows()
```
