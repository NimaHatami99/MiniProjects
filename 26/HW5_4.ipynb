```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv

img = cv.imread('sonography.jpg', 0)
#_______________________________________an implementation 
# show img
plt.figure('output')
plt.subplot(231)
plt.imshow(img, cmap='gray')
plt.title('sono img')

# show sobel
plt.subplot(232)
img_x = cv.filter2D(img, -1, np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]]))
img_y = cv.filter2D(img, -1, np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]))
plt.imshow(abs(img_x) + abs(img_y), cmap='gray', vmin=0, vmax=255)
plt.title('sobel img')

# show prewitt
plt.subplot(2, 3, 3)
img_x = cv.filter2D(img, -1, np.array([[-1, -1, -1], [0, 0, 0], [1, 1, 1]]))
img_y = cv.filter2D(img, -1, np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]]))
plt.imshow(abs(img_x) + abs(img_y), cmap='gray', vmin=0, vmax=255)
plt.title('prewitt img')

# show log
plt.subplot(2, 3, 4)
log = np.array([[0, 0, -1, 0, 0], [0, -1, -2, -1, 0], [-1, -2, 16, -2, -1], [0, -1, -2, -1, 0], [0, 0, -1, 0, 0]])
s_log = cv.filter2D(img, -1, log)
plt.imshow(s_log, cmap='gray', vmin=0, vmax=255)
plt.title('Log img')

# show canny
plt.subplot(2, 3, 5)
s_c = cv.Canny(img, 100, 200)
plt.imshow(s_c, cmap='gray', vmin=0, vmax=255)
plt.title('canny img')

# show roberts
plt.subplot(2, 3, 6)
img_x = cv.filter2D(img, -1, np.array([[-1, 0], [0, 1]]))
img_y = cv.filter2D(img, -1, np.array([[0, -1], [1, 0]]))
plt.imshow(abs(img_x) + abs(img_y), cmap='gray', vmin=0, vmax=255)
plt.title('roberts img')

plt.show()

#_____________________________________________ another implementation 
#-----------sobel----------------------------------
sobel_x = cv.filter2D(img, -1, np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]]))
sobel_y = cv.filter2D(img, -1, np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]))
sobel = abs(sobel_x) + abs(sobel_y)


#----------prewitt----------------------------------
prewitt_x = cv.filter2D(img, -1, np.array([[-1, -1, -1], [0, 0, 0], [1, 1, 1]]))
prewitt_y= cv.filter2D(img, -1, np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]]))
prewitt = abs(prewitt_x)  +  abs(prewitt_y)


#------------LOG------------------------------------
log = cv.filter2D(img, -1, np.array([[0, 0, -1, 0, 0], [0, -1, -2, -1, 0], [-1, -2, 16, -2, -1], [0, -1, -2, -1, 0], [0, 0, -1, 0, 0]]))


#-----------------canny-----------------------------
canny= cv.Canny(img, 100, 200)


#--------------roberts-----------------------------
roberts_x = cv.filter2D(img, -1, np.array([[-1, 0], [0, 1]]))
roberts_y = cv.filter2D(img, -1, np.array([[0, -1], [1, 0]]))
roberts = abs(roberts_x)  +  abs(roberts_y)

#----------------show----------------------------
plt.subplot(231),plt.imshow(img,cmap='gray')
plt.title('orginal'),plt.xticks([]),plt.yticks([])

plt.subplot(232),plt.imshow(sobel,cmap='gray')
plt.title('sobel'),plt.xticks([]),plt.yticks([])

plt.subplot(233),plt.imshow(prewitt,cmap='gray')
plt.title('prewitt'),plt.xticks([]),plt.yticks([])

plt.subplot(234),plt.imshow(log,cmap='gray')
plt.title('LOG'),plt.xticks([]),plt.yticks([])

plt.subplot(235),plt.imshow(canny,cmap='gray')
plt.title('canny'),plt.xticks([]),plt.yticks([])

plt.subplot(236),plt.imshow(roberts,cmap='gray')
plt.title('roberts'),plt.xticks([]),plt.yticks([])

plt.show()
```
