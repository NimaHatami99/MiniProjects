```python
import cv2 as cv 
import matplotlib.pyplot as plt 

img=cv.imread("noisy_rectangle.png",cv.IMREAD_GRAYSCALE)
kernel = cv.getStructuringElement(shape=cv.MORPH_ELLIPSE, ksize=(15, 15))
erosion = cv.erode(img,kernel, iterations = 2)
dilation = cv.dilate(img,kernel, iterations = 2)

plt.subplot(131),plt.imshow(img, cmap = 'gray',vmin  = 0, vmax = 255)
plt.title(' Image'), plt.xticks([]), plt.yticks([])
plt.subplot(132),plt.imshow(erosion, cmap = 'gray',vmin  = 0, vmax = 255)
plt.title('erosion'), plt.xticks([]), plt.yticks([])
plt.subplot(133),plt.imshow(dilation, cmap = 'gray',vmin  = 0, vmax = 255)
plt.title('dilation'), plt.xticks([]), plt.yticks([])

kernel2 = cv.getStructuringElement(cv.MORPH_RECT,(38 , 45))
opening = cv.morphologyEx(img, cv.MORPH_OPEN, kernel2)
kernel3 = cv.getStructuringElement(cv.MORPH_RECT,(19 , 35))
closing = cv.morphologyEx(opening , cv.MORPH_CLOSE, kernel3)

plt.figure()
plt.subplot(121),plt.imshow(opening, cmap = 'gray',vmin  = 0, vmax = 255)
plt.title('opening'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(closing, cmap = 'gray',vmin  = 0, vmax = 255)
plt.title('closing'), plt.xticks([]), plt.yticks([])

plt.show()
```
