```python
import numpy as np
import cv2

img = cv2.imread('reflections.jpg', 0)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        if img[i,j]<50:
            img[i,j]=0
        else:
            img[i,j]=255

def hole_filling(event, y, x, flags, param):
    global img
    if event == cv2.EVENT_LBUTTONDOWN:
        global img
        comp_img = 255 - img
        filled = 255 * (np.ones(img.shape))
        filled = filled.astype('uint8')
        copy = filled.copy()

        kernel = cv2.getStructuringElement(cv2.MORPH_CROSS, (3, 3))

        filled[x, y] = 0
        while np.array_equal(filled, copy) == False:
            copy = filled.copy()
            filled = cv2.erode(filled, kernel)
            filled = cv2.bitwise_or(filled, img)

        img = 255 - (cv2.bitwise_and(filled, comp_img))
        cv2.imshow("image", img)

cv2.imshow("image",img)
cv2.namedWindow("image")
while (True):
    cv2.setMouseCallback("image", hole_filling)
    if cv2.waitKey(10) & 0xFF == 27:
        cv2.destroyWindow
        break
```
