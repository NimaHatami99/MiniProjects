# star

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, filters 
from skimage.segmentation import active_contour

# Loading the star image 
img_gray = io.imread('star.png', as_gray=True)

# Creating an initial circular contour around the star 
s = np.linspace(0, 2*np.pi, 400)
r = 0.5 * min(img_gray.shape) + 0.5 * min(img_gray.shape) * np.sin(s)
c = 0.5 * img_gray.shape[1] + 0.5 * img_gray.shape[1] * np.cos(s)
init = np.array([r, c]).T

# Applying Gaussian filter to the grayscale image 
img_smooth = filters.gaussian(img_gray, sigma=3, preserve_range=False)

# Running active contour segmentation with all parameters 
snake = active_contour(img_smooth, init, alpha=0.001, beta=0.1, w_line=0, w_edge=1, gamma=0.01,
                        max_num_iter=5000, convergence=0.1, coordinates='rc')

# Plotting the original image 
fig, ax = plt.subplots(figsize=(7, 7))
ax.imshow(img_gray, cmap=plt.cm.gray)
ax.plot(snake[:, 1], snake[:, 0], '-b', lw=3)
ax.set_xticks([]), ax.set_yticks([])
ax.axis([0, img_gray.shape[1], img_gray.shape[0], 0])

# Saving the result 
plt.savefig('segmented_star.png')

ax.plot(init[:, 1], init[:, 0], '--r', lw=3)

plt.show()
```

# circle 

```python
# Loading the circle image
img_gray = io.imread('circle.jpg', as_gray=True)

# Creating an initial circular contour around the circle 
s = np.linspace(0, 2*np.pi, 400)
scaling_factor = 0.8
r = 0.5 * min(img_gray.shape) + scaling_factor * 0.5 * min(img_gray.shape) * np.sin(s)
c = 0.5 * img_gray.shape[1] + scaling_factor * 0.5 * img_gray.shape[1] * np.cos(s)
init = np.array([r, c]).T 

# Applying Gaussian filter to the grayscale image
img_smooth = filters.gaussian(img_gray, sigma=3, preserve_range=False)

# Running active contour segmentation with adjusted parameters
snake = active_contour(img_smooth, init, boundary_condition='free',
                       alpha=0.001, beta=0.1, w_line=0, w_edge=1, gamma=0.1,
                       max_num_iter=10000, convergence=0.1, coordinates='rc')

# Plotting the original image
fig, ax = plt.subplots(figsize=(7, 7))
ax.imshow(img_gray, cmap=plt.cm.gray)
ax.plot(snake[:, 1], snake[:, 0], '-b', lw=3)
ax.set_xticks([]), ax.set_yticks([])
ax.axis([0, img_gray.shape[1], img_gray.shape[0], 0])

# Saving the result before displaying
plt.savefig('segmented_circle.jpg')

# Plotting the initial straight line
ax.plot(init[:, 1], init[:, 0], '--r', lw=3)

plt.show()
```
